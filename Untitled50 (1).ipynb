```python
%matplotlib inline

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    auc
)

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

from xgboost import XGBClassifier

from imblearn.over_sampling import RandomOverSampler
```

```python

def bankers_algorithm(allocation, maximum, available):

    allocation = np.array(allocation)
    maximum = np.array(maximum)
    available = np.array(available)

    processes = len(allocation)

    need = maximum - allocation

    finish = [False] * processes

    work = available.copy()

    safe_sequence = []

    while True:

        found = False

        for i in range(processes):

            if not finish[i]:

                if np.all(need[i] <= work):

                    work += allocation[i]

                    finish[i] = True

                    safe_sequence.append(i)

                    found = True

        if not found:
            break

    safe = all(finish)

    return safe, safe_sequence
```

```python

def generate_random_system():

    processes = random.randint(3, 8)

    resources = random.randint(3, 6)

    allocation = np.random.randint(
        0,
        5,
        (processes, resources)
    )

    maximum = allocation + np.random.randint(
        0,
        6,
        (processes, resources)
    )

    total_resources = np.random.randint(
        10,
        30,
        size=(resources,)
    )

    allocated_sum = allocation.sum(axis=0)

    available = total_resources - allocated_sum

    available = np.maximum(available, 0)

    return (
        processes,
        resources,
        allocation,
        maximum,
        available
    )
```

```python
def generate_deadlock_system():

    processes = random.randint(3, 6)

    resources = random.randint(3, 5)

    allocation = np.zeros((processes, resources), dtype=int)

    maximum = np.zeros((processes, resources), dtype=int)

    for i in range(processes):

        allocation[i][i % resources] = 1

        maximum[i][i % resources] = 1

        maximum[i][(i + 1) % resources] = 1

    available = np.zeros(resources, dtype=int)

    return (
        processes,
        resources,
        allocation,
        maximum,
        available
    )
```

```python
def extract_features(
    processes,
    resources,
    allocation,
    maximum,
    available
):

    allocation = np.array(allocation)

    maximum = np.array(maximum)

    available = np.array(available)

    need = maximum - allocation

    total_allocated = np.sum(allocation)

    total_need = np.sum(need)

    total_available = np.sum(available)

    features = {

        'processes': processes,

        'resources': resources,

        'total_allocated': total_allocated,

        'total_need': total_need,

        'total_available': total_available,

        'avg_need': np.mean(need),

        'max_need': np.max(need),

        'min_need': np.min(need),

        'avg_alloc': np.mean(allocation),

        'max_alloc': np.max(allocation),

        'need_variance': np.var(need),

        'alloc_variance': np.var(allocation),

        'resource_pressure': total_need / (total_available + 1),

        'unsafe_ratio': np.sum(need > available) / need.size,

        'need_sum_std': np.std(np.sum(need, axis=1)),

        'alloc_sum_std': np.std(np.sum(allocation, axis=1)),

        'available_mean': np.mean(available),

        'available_min': np.min(available),

        'available_max': np.max(available)
    }

    return features
```

```python
rows = []

print('Generating dataset...')

for _ in range(30000):

    if random.random() < 0.5:

        (
            processes,
            resources,
            allocation,
            maximum,
            available

        ) = generate_random_system()

    else:

        (
            processes,
            resources,
            allocation,
            maximum,
            available

        ) = generate_deadlock_system()

    safe, sequence = bankers_algorithm(
        allocation,
        maximum,
        available
    )

    deadlock = 0 if safe else 1

    features = extract_features(
        processes,
        resources,
        allocation,
        maximum,
        available
    )

    features['deadlock'] = deadlock

    rows.append(features)

df = pd.DataFrame(rows)

print('\nDataset Shape:')
print(df.shape)

print('\nClass Distribution Before Balancing:')
print(df['deadlock'].value_counts())
```

```python

X = df.drop('deadlock', axis=1)

y = df['deadlock']

ros = RandomOverSampler(random_state=42)

X_balanced, y_balanced = ros.fit_resample(X, y)

print('\nClass Distribution After Balancing:')
print(pd.Series(y_balanced).value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_balanced,
    y_balanced,
    test_size=0.2,
    random_state=42,
    stratify=y_balanced
)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)

joblib.dump(scaler, 'scaler.pkl')
```

```python

models = {

    'Logistic Regression': LogisticRegression(
        max_iter=2000,
        class_weight='balanced'
    ),

    'Decision Tree': DecisionTreeClassifier(
        max_depth=12,
        class_weight='balanced'
    ),

    'Random Forest': RandomForestClassifier(
        n_estimators=200,
        max_depth=15,
        class_weight='balanced',
        random_state=42
    ),

    'KNN': KNeighborsClassifier(
        n_neighbors=5
    ),

    'XGBoost': XGBClassifier(
        n_estimators=300,
        max_depth=8,
        learning_rate=0.05,
        subsample=0.9,
        colsample_bytree=0.9,
        eval_metric='logloss',
        random_state=42
    )
}
```

```python
trained_models = {}

results = []

for name, model in models.items():

    print('\n===================================')

    print(f'Training {name}...')

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    print(f'Accuracy: {accuracy:.4f}')

    print('\nClassification Report:')

    print(classification_report(y_test, y_pred))

    trained_models[name] = model

    results.append({
        'Model': name,
        'Accuracy': accuracy
    })
```

```python
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by='Accuracy',
    ascending=False
)

print('\nFinal Results:')
print(results_df)

best_model_name = results_df.iloc[0]['Model']

best_model = trained_models[best_model_name]

print(f'\nBest Model: {best_model_name}')

joblib.dump(best_model, 'best_deadlock_model.pkl')
```

```python
plt.figure(figsize=(10,8))

for name, model in trained_models.items():

    y_prob = model.predict_proba(X_test)[:,1]

    fpr, tpr, _ = roc_curve(y_test, y_prob)

    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        label=f'{name} AUC={roc_auc:.3f}'
    )

plt.plot([0,1],[0,1],'--')

plt.xlabel('False Positive Rate')

plt.ylabel('True Positive Rate')

plt.title('ROC Curve')

plt.legend()

plt.grid()

plt.show()
```

```python
#mano
from IPython.display import HTML
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from matplotlib.patches import FancyArrowPatch
import numpy as np

# =========================
# DATA (لازم تكون معرفة قبل الكود)
# allocation = ...
# maximum = ...
# processes = ...
# resources = ...
# prediction = 0 أو 1
# =========================

allocation_np = np.array(allocation)
maximum_np = np.array(maximum)

process_nodes = [f"P{i}" for i in range(processes)]
resource_nodes = [f"R{i}" for i in range(resources)]

# =========================
# GRAPH
# =========================

G = nx.DiGraph()
G.add_nodes_from(process_nodes + resource_nodes)


#basmalla
edges = []

# Need edges (P -> R)
for i in range(processes):
    for j in range(resources):
        if maximum_np[i][j] - allocation_np[i][j] > 0:
            edges.append((f"P{i}", f"R{j}", "red"))

# Allocation edges (R -> P)
for i in range(processes):
    for j in range(resources):
        if allocation_np[i][j] > 0:
            edges.append((f"R{j}", f"P{i}", "green"))

# =========================
# POSITION (مسافة بين الجانبين فقط)
# =========================
#nour
pos = {}

x_gap = 4   # بين اليمين والشمال
y_gap = 3   # بين العناصر داخل نفس العمود

for i, p in enumerate(process_nodes):
    pos[p] = (x_gap, -i * y_gap)

for j, r in enumerate(resource_nodes):
    pos[r] = (-x_gap, -j * y_gap)

# =========================
# FIGURE
# =========================

fig, ax = plt.subplots(figsize=(16, 9))

arrows = []

# =========================
# NODES
# =========================

def draw_nodes():
    nx.draw_networkx_nodes(G, pos, nodelist=process_nodes,
                           node_color="red", node_size=2000, ax=ax)

    nx.draw_networkx_nodes(G, pos, nodelist=resource_nodes,
                           node_color="orange", node_shape="s",
                           node_size=2000, ax=ax)

    nx.draw_networkx_labels(G, pos, ax=ax, font_size=10)

# =========================
# ARROW
# =========================
#habiba
def get_rad(i):
    return ((i % 5) - 2) * 0.15  # منع التداخل

def draw_arrow(u, v, t, color, rad, i):

    x1, y1 = pos[u]
    x2, y2 = pos[v]

    xm = x1 + (x2 - x1) * t
    ym = y1 + (y2 - y1) * t

    y_offset = 0.1 * (i % 2)

    arrow = FancyArrowPatch(
        (x1, y1),
        (xm, ym + y_offset),
        arrowstyle='->',
        mutation_scale=15,
        color=color,
        lw=2,
        connectionstyle=f"arc3,rad={rad}"
    )

    ax.add_patch(arrow)
    return arrow

# =========================
# UPDATE
# =========================
#shefo
def update(frame):
    global arrows

    for a in arrows:
        a.remove()
    arrows = []

    ax.clear()
    ax.set_axis_off()

    ax.set_xlim(-6, 6)
    ax.set_ylim(-processes * y_gap - 2, resources * y_gap + 2)

    draw_nodes()

    ax.set_title(
        "DEADLOCK DETECTED" if prediction == 1 else "SAFE STATE",
        fontsize=18,
        color="red" if prediction == 1 else "green",
        fontweight="bold"
    )

    step = 10
    idx = frame // step

    # completed edges
    for i in range(min(idx, len(edges))):
        u, v, c = edges[i]
        arrows.append(draw_arrow(u, v, 1, c, get_rad(i), i))

    # moving edge
    if idx < len(edges):
        u, v, c = edges[idx]
        t = (frame % step) / step
        arrows.append(draw_arrow(u, v, t, c, get_rad(idx), idx))

    return arrows

# =========================
# RUN
# =========================

ani = animation.FuncAnimation(
    fig,
    update,
    frames=200,
    interval=120,
    repeat=False
)

plt.close()
HTML(ani.to_jshtml())
```
